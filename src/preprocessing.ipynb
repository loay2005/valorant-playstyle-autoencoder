```python
"""
preprocessing.py
Cleans and prepares the VCT player stats dataset for the Autoencoder pipeline.
"""

import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
def load_data():
    df = pd.read_csv(r'C:\Users\User\Downloads\valorant vct players real.csv')
    return df
```

```python
def clean_data(df):
    percent_cols = ["KAST", "Opening Duel Win %"]
    for col in percent_cols:
        if df[col].dtype == object:
            df[col] = df[col].str.replace("%", "", regex=False).astype(float)
    df = df.dropna()
    return df
```

```python
def select_features(df):
    feature_cols = [
        "ACS", "K:D", "KAST", "ADR", "KPR", "APR",
        "FKPR", "FDPR", "HS%", "CL%", "Opening Duel Win %"
    ]
    return df[feature_cols], df[["Player", "Team", "Region"]]
```

```python
def scale_features(X):
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    return X_scaled, scaler
```

```python
df = load_data()
df = clean_data(df)
X, meta = select_features(df)
X_scaled, scaler = scale_features(X)
print("Data shape after preprocessing:", X_scaled.shape)
meta.head()
```
