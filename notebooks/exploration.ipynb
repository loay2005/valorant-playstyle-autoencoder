```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv(r'C:\Users\User\Downloads\valorant vct players real.csv')
```

```python
df["KAST"] = pd.to_numeric(
    df["KAST"].astype(str).str.replace("%", "", regex=False),
    errors="coerce"
)

df["Opening Duel Win %"] = pd.to_numeric(
    df["Opening Duel Win %"].astype(str).str.replace("%", "", regex=False),
    errors="coerce"
)
```

```python
sns.countplot(data = df , x="Region")
plt.title("Players per Region")
plt.show()
```

```python
sns.histplot(df["ACS"] , kde = True )
plt.title("Distrbution of ACS ( Average Combat score)")
plt.show()
```

```python
features = ["ACS","K:D","KAST","ADR","KPR","APR","FKPR","FDPR","HS%","CL%","Opening Duel Win %"]
corr = df[features].corr()
plt.Figure(figsize=(10, 8))
sns.heatmap(corr , annot=True , cmap="coolwarm", fmt=".2f")
plt.title("Feature Correlation Heatmap")
plt.show
```

```python
sns.scatterplot(data=df , x="FKPR", y="Opening Duel Win %", hue = "Region") 
plt.title("First kill Rate vs Opening Duel Win %") 
plt.show()
```
